```python
import pandas as pd
```

```python
# The dataset taken from Kaggle
imdb_df = pd.read_csv("movie-data/Imdb_Movie_Dataset.csv")
print(imdb_df.info())
print(imdb_df.head(5))
```

```python
# Dataset of principal people IDs.
principals_df = pd.read_csv("movie-data/imdb-pre-processed/title.principals.tsv", sep="\t", dtype=str)
principals_df.drop(columns=["ordering","characters"], inplace=True)
print(principals_df.info())
print(principals_df.head(5))
```

```python
# Maybe use more filtering methods, like try taking only the director, producer, and cinematographer.
```

```python
# Dataset of the names associated with name IDs.
names_df = pd.read_csv("movie-data/imdb-pre-processed/name.basics.tsv", sep="\t", dtype=str)
names_df.drop(columns=["birthYear","deathYear","primaryProfession","knownForTitles"], inplace=True)
print(names_df.info())
print(names_df.head(5))
```

```python
# This creates a new dataset which contains a movie title id, the name of a person associated with it, their name id, category, and job.
full_names_df = pd.merge(principals_df, names_df, on="nconst", how="left")
print(full_names_df.info())
print(full_names_df.head(5))
```

```python

combined_df = pd.merge(imdb_df, full_names_df, left_on="imdb_id", right_on="tconst", how="left")
print(combined_df.info())
print(combined_df.head(25).to_string())
```

```python
combined_df.to_csv('movie-data/combined_df.csv', index=False)
```
