```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.neighbors import KNeighborsRegressor
from impute import impute_zero
```

```python
df = pd.read_csv('movie-data/Imdb_Movie_Dataset.csv')
```

```python
cols = df.columns
```

```python
numeric_columns = ["id","vote_average","vote_count","revenue","runtime","budget","popularity"]
```

```python
categorical_columns = list(set(cols).symmetric_difference(set(numeric_columns)))
```

```python
df_cate = df[[*categorical_columns]]
```
